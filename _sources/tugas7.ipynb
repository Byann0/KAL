Rumus Determinan :

$$\text{det}(A) = \sum_{k=1}^{n} (-1)^{i+k} a_{ik} M_{ik}$$

## Jawab

1. $$A = \begin{bmatrix} -7 & -5 \\ 1 & 4 \end{bmatrix}$$

Ekspansi Baris 1 ($i=1$):
- Elemen $a_{11} = -7$: Minor $M_{11} = |4| = 4$.
- Elemen $a_{12} = -5$: Minor $M_{12} = |1| = 1$.

Perhitungan :

$$\text{det}(A) = a_{11}(-1)^{1+1}M_{11} + a_{12}(-1)^{1+2}M_{12}$$
$$\text{det}(A) = (-7)(1)(4) + (-5)(-1)(1)$$
$$\text{det}(A) = -28 + 5 = \mathbf{-23}$$

2.  $$A = \begin{bmatrix} 0 & 2 & -3 \\ 1 & -2 & -1 \\ 0 & 0 & 1 \end{bmatrix}$$

Ekspansi Baris 3 ($i=3$):
- Elemen $a_{31} = 0$: 
- Elemen $a_{32} = 0$: 
- Elemen $a_{33} = 1$: Minor $M_{33} = \text{det}\begin{bmatrix} 0 & 2 \\ 1 & -2 \end{bmatrix} = (0 \cdot -2) - (2 \cdot 1) = -2$.

Perhitungan :

$$\text{det}(A) = a_{31}C_{31} + a_{32}C_{32} + a_{33}(-1)^{3+3}M_{33}$$
$$\text{det}(A) = 0 + 0 + (1)(1)(-2)$$
$$\text{det}(A) = \mathbf{-2}$$

3.  $$A = \begin{bmatrix} 1 & -3 & 1 & 1 \\ -3 & 1 & 1 & 1 \\ 1 & 1 & -3 & 1 \\ 1 & 1 & 1 & -3 \end{bmatrix}$$

Ekspansi Baris 1 ($i=1$):

$\text{det}(A) = a_{11}C_{11} + a_{12}C_{12} + a_{13}C_{13} + a_{14}C_{14}$

Kofaktor $C_{1k}$:
- $C_{11}$: $(-1)^2 \text{det} \begin{bmatrix} 1 & 1 & 1 \\ 1 & -3 & 1 \\ 1 & 1 & -3 \end{bmatrix} = 1 \cdot [1(9-1) - 1(-3-1) + 1(1+3)] = 8 + 4 + 4 = 16$
- $C_{12}$: $(-1)^3 \text{det} \begin{bmatrix} -3 & 1 & 1 \\ 1 & -3 & 1 \\ 1 & 1 & -3 \end{bmatrix} = -1 \cdot [-3(9-1) - 1(-3-1) + 1(1+3)] = -1[-24+4+4] = 16$
- $C_{13}$: $(-1)^4 \text{det} \begin{bmatrix} -3 & 1 & 1 \\ 1 & 1 & 1 \\ 1 & 1 & -3 \end{bmatrix} = 1 \cdot [-3(-3-1) - 1(-3-1) + 1(1-1)] = 12 + 4 + 0 = 16$
- $C_{14}$: $(-1)^5 \text{det} \begin{bmatrix} -3 & 1 & 1 \\ 1 & 1 & -3 \\ 1 & 1 & 1 \end{bmatrix} = -1 \cdot [-3(1+3) - 1(1+3) + 1(1-1)] = -1[-12-4] = 16$

Perhitungan :

$$\text{det}(A) = (1)(16) + (-3)(16) + (1)(16) + (1)(16)$$
$$\text{det}(A) = 16 - 48 + 16 + 16$$
$$\text{det}(A) = 48 - 48 = \mathbf{0}$$

B. Gunakan rumus matriks adjoin untuk menghitung invers dari matriks berikut dengan rumus

Rumus Matriks Adjoin :

$$A^{-1} = \frac{1}{\text{det}(A)} \text{adj}(A)$$

4.  $$A = \begin{bmatrix} -7 & -5 \\ 1 & 4 \end{bmatrix}$$

Determinan :

$\text{det}(A) = -23$.

Matriks Adjoin :

$$\text{adj}(A) = \begin{bmatrix} 4 & 5 \\ -1 & -7 \end{bmatrix}$$

Invers :

$$A^{-1} = \frac{1}{-23} \begin{bmatrix} 4 & 5 \\ -1 & -7 \end{bmatrix} = \mathbf{\begin{bmatrix} -4/23 & -5/23 \\ 1/23 & 7/23 \end{bmatrix}}$$

5.  $$A = \begin{bmatrix} 0 & 2 & -3 \\ 1 & -2 & -1 \\ 0 & 0 & 1 \end{bmatrix}$$

Determinan :

$\text{det}(A) = -2$.

Matriks Kofaktor :
Kofaktor ($C_{ij}$):

- $C_{11} = + |_{-2 \ -1}^{ \ 0 \ \ \ 1}| = -2$
- $C_{12} = - |^{1 \ -1}_{0 \ \ \ 1}| = -1$
- $C_{13} = + |^{1 \ -2}_{0 \ \ \ 0}| = 0$
- $C_{21} = - |^{2 \ -3}_{0 \ \ \ 1}| = -2$
- $C_{22} = + |^{0 \ -3}_{0 \ \ \ 1}| = 0$
- $C_{23} = - |^{0 \ \ \ 2}_{0 \ \ \ 0}| = 0$
- $C_{31} = + |^{ \ 2 \ -3}_{-2 \ -1}| = -2 - 6 = -8$
- $C_{32} = - |^{0 \ -3}_{1 \ -1}| = -(0 - (-3)) = -3$
- $C_{33} = + |^{0 \ \ \ 2}_{1 \ -2}| = 0 - 2 = -2$

Adjoin(Tranpose) :

$$\text{adj}(A) = \begin{bmatrix} -2 & -1 & 0 \\ -2 & 0 & 0 \\ -8 & -3 & -2 \end{bmatrix}^T$$

$$\text{adj}(A) = K^T = \begin{bmatrix} -2 & -2 & -8 \\ -1 & 0 & -3 \\ 0 & 0 & -2 \end{bmatrix}$$

Invers :

$$A^{-1} = \frac{1}{-2} \begin{bmatrix} -2 & -2 & -8 \\ -1 & 0 & -3 \\ 0 & 0 & -2 \end{bmatrix} = \mathbf{\begin{bmatrix} 1 & 1 & 4 \\ 0.5 & 0 & 1.5 \\ 0 & 0 & 1 \end{bmatrix}}$$
